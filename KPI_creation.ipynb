```python
# Create widgets for the SQL query parameters
# catalog / schema / table for IDENTIFIER() in the next cell
dbutils.widgets.text("catalog_name", "workspace.auto_loader", "Catalog name")
dbutils.widgets.text("schema_name", "demo", "Schema name")
dbutils.widgets.text("table_name", "demo_table2", "Table name")



# Read widget values into Python variables
widget_catalog = getArgument("catalog_name")
widget_schema = getArgument("schema_name")
widget_table = getArgument("table_name")


print(f"Catalog : {widget_catalog}")
print(f"Schema  : {widget_schema}")
print(f"Table   : {widget_table}")


# Build the fully-qualified table name
full_table = f"{widget_catalog}.{widget_schema}.{widget_table}"
print(f"\nFull table: {full_table}")
```
