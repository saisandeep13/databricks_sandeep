```python
df = spark.sql("select * from workspace.bronze1.coffesales")
display(df)
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import sum, avg, round

window_spec = Window.partitionBy("cash_type")

df_window = df.withColumn("total_sales_by_cash_type", round(sum("money").over(window_spec), 3)) \
              .withColumn("avg_sales_by_cash_type", round(avg("money").over(window_spec), 3))

display(df_window)
```

```python
from pyspark.sql.functions import count

df_window = df_window.withColumn("count_cash_type", count("cash_type").over(window_spec))

display(df_window)
```

```python
df_window.write.mode("append").saveAsTable("workspace.silver1.coffesales")
```
