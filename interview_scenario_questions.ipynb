````python
from pyspark.sql.functions import (
    col, initcap, trim, lower, upper, regexp_replace, when, lit, coalesce, expr
)
from pyspark.sql import SparkSession

spark = SparkSession.builder.getOrCreate()

df = spark.read.table("workspace.bronze.employee")

df_standardized = (
    df
    # 1. employee_id — trim & cast to int
    .withColumn("employee_id", trim(col("employee_id")).cast("int"))

    # 2. employee_name — trim & initcap
    .withColumn("employee_name", initcap(trim(col("employee_name"))))

    # 3. department — trim, initcap, blank → null
    .withColumn("department",
        when(trim(col("department")) == "", None)
        .otherwise(upper(trim(col("department")))))

    # 4. salary — strip non-numeric chars & cast to double (null when nothing numeric remains)
    .withColumn("salary",
        expr("try_cast(regexp_replace(salary, '[^0-9.]', '') AS DOUBLE)"))

    # 5. phone_number — keep digits only & cast to string (null if invalid)
    .withColumn("phone_number",
        when(
            regexp_replace(col("phone_number"), r"[^0-9]", "").rlike(r"^\d{10,13}$"),
            regexp_replace(col("phone_number"), r"[^0-9]", "")
        ).otherwise(None))

    # 6. email — trim & lowercase
    .withColumn("email", lower(trim(col("email"))))

    # 7. city — trim, initcap, blank → null
    .withColumn("city",
        when(trim(col("city")) == "", None)
        .otherwise(initcap(trim(col("city")))))

    # 8. state — trim, initcap, blank or 'NA' → null
    .withColumn("state",
        when(trim(col("state")).isin("", "NA"), None)
        .otherwise(initcap(trim(col("state")))))

    # 9. joining_date — try common formats & cast to date
    .withColumn("joining_date",
        coalesce(
            expr("try_to_date(joining_date, 'yyyy-MM-dd')"),
            expr("try_to_date(joining_date, 'dd/MM/yyyy')"),
            expr("try_to_date(joining_date, 'dd-MM-yyyy')"),
            expr("try_to_date(joining_date, 'MM-dd-yyyy')")
        ))

    # 10. employment_status — trim & map to Active / Inactive / null
    .withColumn("employment_status",
        when(upper(trim(col("employment_status"))) == "ACTIVE", lit("Active"))
        .when(upper(trim(col("employment_status"))).isin("INACTIVE", "TERMINATED", "RESIGNED"), lit("Inactive"))
        .otherwise(None))

    # 11. manager_id — trim, blank/0 → null & cast to int
    .withColumn("manager_id",
        when(trim(col("manager_id")).isin("", "0"), None)
        .otherwise(trim(col("manager_id")).cast("int")))

    # 12. remarks — trim & clean up special characters
    .withColumn("remarks", trim(regexp_replace(col("remarks"), r"[^a-zA-Z0-9\s]", "")))
)

display(df_standardized)
````

# Concurrency Control in Delta Lake

## Overview

When multiple users or jobs try to read and write to the **same table** at the same time, the system needs a mechanism to ensure data integrity. This is called **Concurrency Control**.

Delta Lake handles this using two key concepts:

| Concept | What It Does |
| --- | --- |
| **Optimistic Concurrency Control (OCC)** | Assumes conflicts are rare — lets transactions proceed without locking, then validates before committing |
| **Row-Level Concurrency** | Allows concurrent writes to **different rows** in the **same file** without conflict |

The **Delta Transaction Log** (`_delta_log/`) is the backbone that makes both of these work — it records every change as an ordered, immutable history of JSON files.

## 1. Optimistic Concurrency Control (OCC)

### What is OCC?

**Optimistic Concurrency Control** is a strategy where transactions assume that **conflicts will be rare**. Instead of locking data when a transaction starts (pessimistic approach), OCC lets transactions proceed freely and only checks for conflicts **at commit time**.

### OCC vs Pessimistic Concurrency Control

| Aspect | Pessimistic | Optimistic (OCC) |
| --- | --- | --- |
| Locking | Locks data at the **start** | No locks — validates at **commit** |
| Assumption | Conflicts are **frequent** | Conflicts are **rare** |
| Throughput | Lower (blocking) | Higher (non-blocking) |
| Conflict handling | Wait for lock release | **Abort & retry** |
| Best for | High contention | Low contention (common in data lakes) |

### The Three Phases of OCC

1. **Read Phase** — Transaction reads data and makes changes in a private buffer. No locks are acquired.
2. **Validation Phase** — Before committing, the system checks whether any other transaction has modified the data this transaction read or wrote.
3. **Write Phase** — If validation passes, the changes are committed. If validation fails, the transaction is **aborted and must retry**.

### Why OCC for Delta Lake?

Data lake workloads typically have **low contention**:
- Jobs run at different times (batch windows)
- Concurrent jobs usually touch different partitions/files
- Locking would unnecessarily block long-running analytical queries

OCC allows maximum parallelism while guaranteeing correctness.

## 2. How Delta Lake Implements OCC

### Isolation Levels

Delta Lake supports two isolation levels:

| Isolation Level | Description | Default |
| --- | --- | --- |
| **WriteSerializable** | Write operations must be serializable. Reads may see uncommitted writes. | Yes (for writes) |
| **Serializable** | Strongest — all operations (reads + writes) appear to execute sequentially. | Yes (for reads) |

You can set it with:
```sql
ALTER TABLE my_table SET TBLPROPERTIES ('delta.isolationLevel' = 'Serializable');
```

### The OCC Commit Protocol in Delta Lake

When a transaction commits, Delta Lake follows this flow:

```
Transaction starts → reads table at version N → makes changes in buffer
  → tries to commit as version N+1
  → checks: has the table moved past version N?
     ├─ No  → commit succeeds (writes log file N+1)
     └─ Yes → conflict resolution:
        ├─ Are the files I modified different from what others modified?
        │   ├─ Yes (no file overlap) → commit STILL succeeds
        │   └─ No (file overlap) → ConcurrentModificationException → ABORT
```

### Key Insight: File-Level Conflict Detection

Delta Lake does **not** lock the entire table. Instead, it checks at the **file level**:
- Each transaction records which files it read and which files it added/removed
- At commit time, it compares against all transactions that committed in the meantime
- If the **same data file** was modified by both, it's a conflict
- If different files were touched, both transactions can safely coexist

This means two concurrent UPDATEs on **different partitions** will **never conflict**, even without locking.

## 3. Row-Level Concurrency

### What is Row-Level Concurrency?

Without row-level concurrency, Delta Lake detects conflicts at the **file level** — if two transactions modify the same Parquet file (even different rows), one will fail.

**Row-level concurrency** goes a step further: it allows two transactions to modify **different rows within the same file** without conflict.

### Before vs After Row-Level Concurrency

```
WITHOUT row-level concurrency:
  Transaction A updates row 1 in file_1.parquet
  Transaction B updates row 2 in file_1.parquet
  → CONFLICT (same file touched by both)
  → One transaction is aborted

WITH row-level concurrency:
  Transaction A updates row 1 in file_1.parquet
  Transaction B updates row 2 in file_1.parquet
  → NO CONFLICT (different rows in the same file)
  → Both transactions succeed!
```

### How It Works Under the Hood

1. **MERGE, UPDATE, DELETE** operations track which **specific rows** (by row position / file offset) were modified
2. At commit time, Delta compares the **row-level modification ranges** between concurrent transactions
3. If the row ranges don't overlap, the commit proceeds — Delta merges both changes into a new version of the file
4. If the row ranges **do** overlap, it's a genuine conflict → abort

### When Is It Useful?

- **Concurrent MERGE operations** from multiple streaming jobs writing to the same target table
- **Multi-threaded CDC** (Change Data Capture) where different threads update different rows
- **High-frequency upserts** where many small updates hit the same file

### Requirements

Row-level concurrency is automatically used when:
- The table uses Delta Lake (it's a built-in feature)
- Operations are `MERGE`, `UPDATE`, or `DELETE` (not plain `INSERT`)
- The conflicting transactions modify **different rows** in the same file

## 4. How the Delta Transaction Log Handles Concurrency

### The `_delta_log/` Directory

Every Delta table has a `_delta_log/` folder at its root. This is the **single source of truth** for the table's state.

```
workspace.bronze.employee/
├── _delta_log/
│   ├── 00000000000000000000.json   ← Initial commit (CREATE TABLE)
│   ├── 00000000000000000001.json   ← Second transaction
│   ├── 00000000000000000002.json   ← Third transaction
│   ├── ...
│   ├── 00000000000000000010.checkpoint.parquet   ← Checkpoint (every N commits)
│   └── _last_checkpoint             ← Metadata about the latest checkpoint
├── part-00000-xxx.snappy.parquet     ← Actual data files
├── part-00001-yyy.snappy.parquet
└── ...
```

### How a Commit Works (Atomic & Ordered)

1. Transaction reads the current table state (e.g., at version 5)
2. Transaction makes changes and prepares a new JSON log entry
3. To commit, it tries to **create file `00000000000000000006.json`**
4. The cloud storage (S3, ADLS, GCS) provides **atomic create-if-not-exists**:
   - If the file **doesn't exist** → create succeeds → commit succeeds (table is now at version 6)
   - If the file **already exists** → create fails → **another transaction won** → conflict!

This is the core of OCC in Delta Lake: **the atomic file creation IS the lock**. No separate locking mechanism is needed.

### What's Inside a Delta Log JSON File?

Each JSON file contains a list of **actions**:

```json
{"add": {"path": "part-00000-xxx.parquet", "partitionValues": {}, "size": 1234, "modificationTime": 1633000000}}
{"remove": {"path": "part-00000-old.parquet", "deletionTimestamp": 1633000001}}
{"commitInfo": {"operation": "WRITE", "operationParameters": {...}, "isolationLevel": "WriteSerializable"}}
```

| Action | Purpose |
| --- | --- |
| `add` | A new data file was added |
| `remove` | An existing data file was logically removed |
| `commitInfo` | Metadata about the transaction (operation, user, time, isolation level) |
| `metaData` | Table schema, configuration, partition columns |
| `protocol` | Min reader/writer protocol version |
| `txn` | Transaction app-id tracking (for idempotent writes) |

### Checkpointing — Why and How

After many commits, replaying all JSON files to reconstruct the table state becomes slow. Delta Lake creates **checkpoints**:

- Every **10 commits** (configurable via `delta.checkpointInterval`), a `.checkpoint.parquet` file is written
- It contains the **complete current state** (all active `add` actions, schema, etc.)
- Readers start from the latest checkpoint and only replay the few JSON files after it

```
Checkpoint at v10:  00000000000000000010.checkpoint.parquet
                     00000000000000000011.json
                     00000000000000000012.json
Reader starts at checkpoint v10, then replays v11, v12 → gets current state
```

### How This Enables OCC End-to-End

```

  TXN-A reads table at version 5        TXN-B reads table at version 5
          │                                    │
          ▼                                    ▼
  Makes changes in buffer             Makes changes in buffer
  (no locks!)                         (no locks!)
          │                                    │
          ▼                                    ▼
  Tries to create v6.json             Tries to create v6.json
  ✅ SUCCESS → committed              ❌ FAIL (already exists)
          │                                    │
          ▼                                    ▼
  Table is now at v6                   Conflict resolution:
                                      - Did TXN-B touch same files as TXN-A?
                                        ├─ No  → retry as v7 → SUCCESS
                                        └─ Yes → abort with ConcurrentModificationException
```

   
## Delta Lake Isolation Levels — WriteSerializable, Serializable & Snapshot Isolation

### What Is an Isolation Level?

An **isolation level** defines *what a reader is allowed to see* when another transaction is in the middle of modifying the same table. It is a contract between the database engine and the user about the consistency of reads in the presence of concurrent writes.

Delta Lake supports two isolation levels for its transactions, and underneath both of them it relies on **Snapshot Isolation** as the read mechanism.

---

### 1. WriteSerializable (Default)

`WriteSerializable` is the **default isolation level** for Delta Lake writes.

#### What it guarantees

| Rule | Detail |
| --- | --- |
| **Write serializability** | All **write** operations (INSERT, UPDATE, DELETE, MERGE) are serializable — they appear to execute one after another in some sequential order. |
| **Read consistency** | A reader sees a **consistent snapshot** of the table at the moment the read started. It never sees partial writes. |
| **What it does NOT guarantee** | Read operations do **not** need to be serializable with respect to writes. A long-running read that started *before* a write committed may continue to see the **old** snapshot even after the write is committed. |

#### Why is it the default?

Most data-lake workloads are **append-only** or **partition-disjoint**. Two concurrent writes rarely touch the same files, so full serializability would add overhead without practical benefit. `WriteSerializable` gives the strongest guarantee you actually need (no corrupt writes) while allowing maximum read parallelism.

#### When can a conflict happen?

A `ConcurrentModificationException` is raised only when two concurrent **writes** touch the **same data file** (or the same row, if row-level concurrency is active). A concurrent **read** never blocks a write and never fails.

```
  ┌─────────────────────────────────────────────────────────┐
  │              WriteSerializable                           │
  │                                                         │
  │  Writer A:  UPDATE file_1.parquet  ──┐                  │
  │                                       ├─ same file?     │
  │  Writer B:  UPDATE file_1.parquet  ──┘  YES → CONFLICT  │
  │                                                         │
  │  Writer A:  UPDATE file_1.parquet  ──┐                  │
  │                                       ├─ same file?     │
  │  Writer B:  UPDATE file_2.parquet  ──┘  NO  → both OK    │
  │                                                         │
  │  Reader  R:  SELECT *  ── sees snapshot at read start   │
  │             (never blocked, never fails)                 │
  └─────────────────────────────────────────────────────────┘
```

---

### 2. Serializable (Strongest)

`Serializable` is the **strongest** isolation level Delta Lake offers. It can be set per table:

```sql
ALTER TABLE my_table SET TBLPROPERTIES ('delta.isolationLevel' = 'Serializable');
```

#### What it guarantees

| Rule | Detail |
| --- | --- |
| **Full serializability** | **All operations** — reads AND writes — must appear to execute in some sequential order. |
| **Read-write conflicts** | If a long-running read started at version N, and a write commits at version N+1 that modifies a file the reader is still scanning, Delta may detect a **read-write conflict** and abort the **writer** (not the reader). |
| **Use case** | Workloads where readers must not miss committed changes or see stale data — e.g., compliance reporting, financial reconciliation, or when downstream logic depends on reading the very latest state. |

#### What changes vs WriteSerializable?

The **only** difference is how **readers** interact with **concurrent writers**:

```
  ┌──────────────────────────────────────────────────────────┐
  │                 Serializable                              │
  │                                                          │
  │  Reader  R:  starts long SELECT at version 5              │
  │              reads file_1.parquet (takes 60 seconds)      │
  │                                                          │
  │  Writer  W:  commits UPDATE on file_1.parquet             │
  │              → version 6                                  │
  │              → file_1.parquet is now removed/replaced     │
  │                                                          │
  │  At W's commit time:                                      │
  │    W checks: is any reader still using file_1?            │
  │      ├─ YES → ConcurrentModificationException on W        │
  │      └─ NO  → W commits normally                          │
  └──────────────────────────────────────────────────────────┘
```

Under `WriteSerializable` the writer would **always** commit (readers are ignored); under `Serializable` the writer can be **aborted** if a reader is still depending on the old file.

#### Comparison Table

| Aspect | WriteSerializable | Serializable |
| --- | --- | --- |
| Write-write conflict detection | ✅ Yes | ✅ Yes |
| Read-write conflict detection | ❌ No | ✅ Yes |
| Read sees consistent snapshot | ✅ Always | ✅ Always |
| Read can block a write | ❌ Never | ✅ Can abort the writer |
| Default | ✅ Yes | No (opt-in) |
| Overhead | Lower | Higher (tracks active readers) |
| Best for | Most data-lake workloads | Strict correctness / compliance |

---

### 3. Snapshot Isolation (the underlying read mechanism)

**Snapshot Isolation** is not a Delta-specific property you set — it is the **mechanism** Delta Lake uses for *every* read, regardless of whether the table is `WriteSerializable` or `Serializable`.

#### How it works

1. When a read transaction starts, it looks at the Delta transaction log and picks a **table version** (e.g., version 5).
2. It collects the list of all `add` actions active at version 5 — these are the **data files** that make up the table at that point in time.
3. The reader then scans **only those files**. Even if a writer commits version 6 (adding new files and removing old ones), the reader continues to see the **version-5 snapshot** because it never looks at the new log entries.
4. The reader always sees a **consistent, complete** view — no torn writes, no partial updates.

```
   Delta Transaction Log                         Reader's view
  ┌──────────────────┐
  │ v0  CREATE        │
  │ v1  INSERT        │         Reader starts at v5:
  │ v2  UPDATE        │           ┌──────────────────────────┐
  │ v3  INSERT        │           │ Snapshot @ v5            │
  │ v4  DELETE        │     ────► │  files: f1, f2, f3, f4   │
  │ v5  UPDATE  ◄─────┼─ reader  │  (consistent, frozen)    │
  │ v6  INSERT  ◄─────┼─ writer  └──────────────────────────┘
  │ v7  MERGE         │
  └──────────────────┘           Writer commits v6 (adds f5,
                                 removes f3). Reader still
                                 sees f1,f2,f3,f4 — the v5
                                 snapshot — until it finishes.
```

#### Key properties of Snapshot Isolation

| Property | Explanation |
| --- | --- |
| **No dirty reads** | A reader never sees uncommitted writes. |
| **No non-repeatable reads** | Within one read transaction, re-reading the same row always returns the same value (the snapshot is frozen). |
| **No phantom reads** | New rows inserted by a concurrent writer are invisible (the file list is fixed at read-start). |
| **Writers are not blocked** | A reader holds no locks; writers can proceed in parallel. |
| **Readers are not blocked** | Writers do not wait for readers to finish (under WriteSerializable). |

#### Snapshot Isolation vs the two Delta levels

```
  ┌──────────────────────────────────────────────────────────────┐
  │  Snapshot Isolation  ─── the READ mechanism (always on)       │
  │     │                                                        │
  │     │  Writer commits while a reader holds an old snapshot    │
  │     │                                                        │
  │     ├─► WriteSerializable: writer ALWAYS commits;            │
  │     │   reader keeps its old snapshot — no conflict.          │
  │     │                                                        │
  │     └─► Serializable: writer may be ABORTED if the reader    │
  │         is still using a file the writer modifies.           │
  └──────────────────────────────────────────────────────────────┘
```

---

### End-to-End Timeline Example

```
  Time ──────────────────────────────────────────────────────────►

  T0  Reader-R starts SELECT (table at version 5)
      │  snapshot = {f1, f2, f3}
      │
  T1  Writer-W1 starts UPDATE on f2 (salary += 10%)
      │
  T2  Writer-W2 starts UPDATE on f4 (new hire)
      │  (f4 not in Reader-R's snapshot)
      │
  T3  Writer-W1 commits → version 6
      │  log: remove f2, add f2'
      │
  │  WriteSerializable:  W1 commits OK, Reader-R still sees f2  │
  │  Serializable:       W1 commits OK (R hasn't touched f2 yet) │
      │
  T4  Writer-W2 commits → version 7
      │  log: add f4 (no removal — append)
      │  OK under both levels (f4 not in R's snapshot, no overlap)
      │
  T5  Reader-R finishes — saw only {f1, f2, f3} (version 5)      │
      │  Never saw f2' or f4 — consistent, stale-but-correct view │
      │
  T6  Reader-R2 starts SELECT (table now at version 7)           │
      │  snapshot = {f1, f2', f3, f4} — sees both committed writes │
  ──────────────────────────────────────────────────────────────
```

### Summary

| Concept | One-liner |
| --- | --- |
| **WriteSerializable** | Writes are serializable; readers always get a consistent snapshot but never interfere with writers. Default. |
| **Serializable** | Everything (reads + writes) is serializable; a writer can be aborted if it modifies a file an active reader is still using. |
| **Snapshot Isolation** | The mechanism powering both levels — every read sees a frozen, consistent table version from the Delta log. |

````python
# ── Simulate OCC Conflict: Two concurrent updates to the SAME file/row ──
# Thread 1: Update Alice's salary to 1100
# Thread 2: Update Alice's salary to 1200
# Both read the table at version 0, but only one can create version 1.

from concurrent.futures import ThreadPoolExecutor, as_completed
from delta.tables import DeltaTable

def update_salary(emp_id, new_salary, thread_name):
    """Attempt to update a single employee's salary."""
    spark_local = SparkSession.builder.getOrCreate()
    try:
        dt = DeltaTable.forName(spark_local, "workspace.bronze.occ_demo")
        dt.update(
            condition=col("id") == emp_id,
            set={"salary": lit(new_salary)}
        )
        return f"{thread_name}: SUCCESS — set id={emp_id} salary={new_salary}"
    except Exception as e:
        return f"{thread_name}: FAILED — {type(e).__name__}: {str(e)[:200]}"

# Reset the table to a known state
spark.sql("""
CREATE OR REPLACE TABLE workspace.bronze.occ_demo
USING DELTA
AS SELECT * FROM VALUES
  (1, 'Alice',   1000),
  (2, 'Bob',     2000),
  (3, 'Charlie', 3000),
  (4, 'David',   4000),
  (5, 'Eve',     5000)
  AS t(id, name, salary)
""")

print("=== Concurrent update to the SAME row (id=1) — OCC conflict scenario ===\n")

results = []
with ThreadPoolExecutor(max_workers=2) as executor:
    futures = [
        executor.submit(update_salary, 1, 1100, "Thread-A"),
        executor.submit(update_salary, 1, 1200, "Thread-B"),
    ]
    for future in as_completed(futures):
        results.append(future.result())

# Sort results for deterministic display
for r in sorted(results):
    print(r)

print("\nTable after concurrent updates:")
display(spark.read.table("workspace.bronze.occ_demo"))
print("\nNote: One thread won the race. The other either retried and succeeded,\nor got a ConcurrentModificationException.")
````

````python
# ── Demonstrate non-conflicting concurrent writes (different rows / different files) ──
# This shows that OCC allows parallelism when transactions touch DIFFERENT data.

# Reset table
spark.sql("""
CREATE OR REPLACE TABLE workspace.bronze.occ_demo
USING DELTA
AS SELECT * FROM VALUES
  (1, 'Alice',   1000),
  (2, 'Bob',     2000),
  (3, 'Charlie', 3000),
  (4, 'David',   4000),
  (5, 'Eve',     5000)
  AS t(id, name, salary)
""")

print("=== Concurrent updates to DIFFERENT rows — should both succeed ===\n")

results2 = []
with ThreadPoolExecutor(max_workers=2) as executor:
    futures = [
        executor.submit(update_salary, 1, 1100, "Thread-A"),  # Update Alice (id=1)
        executor.submit(update_salary, 3, 3300, "Thread-B"),  # Update Charlie (id=3)
    ]
    for future in as_completed(futures):
        results2.append(future.result())

for r in sorted(results2):
    print(r)

print("\nTable after non-conflicting concurrent updates:")
display(spark.read.table("workspace.bronze.occ_demo"))
print("\nNote: Both updates succeeded because they touched DIFFERENT rows.\nDelta's OCC detected no file-level (or row-level) overlap.")
````

## 5. Summary — Key Takeaways

### The Big Picture

```
Delta Lake Concurrency
├── Optimistic Concurrency Control (OCC)
│   ├── Assumes conflicts are rare → no locking during read/processing
│   ├── Validates at commit time → compares file-level changes
│   └── On conflict → ConcurrentModificationException → retry
│
├── Row-Level Concurrency
│   ├── Refines conflict detection from FILE level to ROW level
│   ├── Different rows in the same file → NO conflict
│   ├── Same row in the same file → conflict
│   └── Automatic for MERGE / UPDATE / DELETE operations
│
└── Delta Transaction Log (_delta_log/)
    ├── Ordered JSON files (version 0, 1, 2, ...)
    ├── Atomic commit = atomic file creation in cloud storage
    ├── Each file contains add / remove / commitInfo actions
    ├── Checkpointing every N commits for fast state reconstruction
    └── This log IS the concurrency mechanism — no external lock manager needed
```

### Interview Cheat Sheet

| Question | Answer |
| --- | --- |
| What is OCC? | A concurrency method that doesn't lock — validates at commit, retries on conflict |
| Why does Delta use OCC? | Data lake workloads have low contention; OCC maximizes parallelism |
| What are Delta's isolation levels? | `WriteSerializable` (default for writes) and `Serializable` (strongest) |
| How does a Delta commit work? | Tries to create `N+1.json` atomically; if it exists, conflict resolution kicks in |
| What is row-level concurrency? | Allows concurrent writes to different rows in the same file without conflict |
| What does the delta log contain? | Ordered JSON files with `add`, `remove`, `commitInfo`, `metaData`, `protocol` actions |
| What is checkpointing? | A Parquet snapshot of table state written every ~10 commits to speed up reads |
| When does a conflict happen? | When two transactions modify the same data file (or same rows with row-level concurrency) |
| What happens on conflict? | `ConcurrentModificationException` — the losing transaction must retry or fail |
