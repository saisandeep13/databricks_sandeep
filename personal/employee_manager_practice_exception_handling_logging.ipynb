```python
import uuid
from pyspark.sql import DataFrame
from pyspark.sql import functions as F

df_manager = spark.read.table("workspace.bronze.manager_table")
run_id = str(uuid.uuid4())

def manager_table_columns(df_manager: DataFrame, run_id: str) -> DataFrame:
    return (
        df_manager
        .withColumn('silver_processed_at', F.current_timestamp())
        .withColumn('silver_ingested_at',  F.current_timestamp())
        .withColumn('silver_run_id',       F.lit(run_id))
    )

df_manager_silver = manager_table_columns(df_manager, run_id)
display(df_manager_silver)
```

```python
df_employee = spark.read.table("workspace.bronze.employee_table")
display(df_employee)
```

```python
df_manager = spark.read.table("workspace.bronze.manager_table")
display(df_manager)
```

```python
from pyspark.sql import functions as F

df_join = (df_employee
    .join(df_manager.select(
        "manager_id", "manager_name", "designation",
        F.col("salary").alias("manager_salary"),
        F.col("department").alias("manager_department")),
        df_employee["manager_id"] == df_manager["manager_id"],"left"))
display(df_join) 
```

```python
from pyspark.sql import functions as F

df_grouped = (df_join
    .groupBy(df_manager["manager_id"], "manager_name", "designation")
    .agg(
        F.count("employee_id").alias("report_count"),
        F.sum("salary").alias("total_team_salary"),
        F.round(F.avg("salary")).alias("avg_team_salary"),
        F.max("salary").alias("max_team_salary"),
        F.min("salary").alias("min_team_salary"),
        F.collect_list("employee_name").alias("team_members"),
        F.to_json(F.collect_list("employee_name")).alias("team_members_json"),
        F.split("team_members_json",",").alias("split_team_members_json"),
        F.collect_list("employee_name").alias("team_members_json_array")
    )
    .orderBy("manager_id"))

display(df_grouped)
```

```python
from pyspark.sql import functions as F
from pyspark.sql.types import StructType, StructField, IntegerType, StringType

schema = StructType([
    StructField("paragraph_number", IntegerType(), True),
    StructField("paragraph_text", StringType(), True),
])

data = [
    (1, "Data engineering involves collecting, transforming, and storing large volumes of information so that teams can analyze it for better business decisions daily. "
     "Apache Spark is a powerful engine that processes big data quickly by distributing tasks across multiple nodes to achieve high performance, scalability, and reliability. "
     "Delta Lake adds reliability features like ACID transactions and schema enforcement to data lakes, making them suitable for production pipelines and enterprise environments."),
    (2, "Cloud computing has revolutionized the way organizations build, deploy, and scale applications by providing on-demand access to shared computing resources over the internet. "
     "Containers and orchestration tools like Kubernetes allow teams to package, manage, and automate the deployment of microservices across diverse infrastructure environments efficiently. "
     "Infrastructure as code enables engineers to define, provision, and manage cloud resources using declarative configuration files, ensuring consistency and reproducibility everywhere."),
    (3, "Machine learning models learn patterns from historical data, and they use those patterns to make predictions, classify objects, and generate recommendations for end users. "
     "Feature engineering is a critical step that involves selecting, transforming, and creating input variables to improve the accuracy and performance of ML models. "
     "Model evaluation metrics such as precision, recall, and F1 score help data scientists assess, compare, and tune algorithms before deploying them to production."),
]

df_paragraphs = spark.createDataFrame(data, schema)
display(df_paragraphs)
```

```python
df_clean_count_method = (df_paragraphs
    .withColumn("paragraph_text", F.trim(F.col("paragraph_text")))
    .withColumn("split", F.split("paragraph_text", " "))
    .withColumn("count",F.array_size("split")))
display(df_clean_count_method)
```

```python
df_clean_explode_method = (df_paragraphs
    .withColumn("paragraph_text", F.trim(F.col("paragraph_text")))
    .withColumn("split", F.explode(F.split("paragraph_text", " ")))
    .groupby("paragraph_number")
    .agg(F.count("*").alias("count_words")))
display(df_clean_explode_method)
```

```python
# Step 1: Persist df_paragraphs as a table so SQL views can reference it
df_paragraphs.write.mode("overwrite").saveAsTable("workspace.bronze.paragraphs_table")

# Step 2: Create a regular VIEW
spark.sql("""
CREATE OR REPLACE VIEW workspace.bronze.vw_paragraph_word_count AS
SELECT paragraph_number, COUNT(*) AS count_words
FROM workspace.bronze.paragraphs_table
LATERAL VIEW EXPLODE(SPLIT(TRIM(paragraph_text), ' ')) AS word
GROUP BY paragraph_number
""")

# # Step 3: Create a MATERIALIZED VIEW
# spark.sql("""
# CREATE OR REPLACE MATERIALIZED VIEW workspace.bronze.mv_paragraph_word_count AS
# SELECT paragraph_number, COUNT(*) AS count_words
# FROM workspace.bronze.paragraphs_table
# LATERAL VIEW EXPLODE(SPLIT(TRIM(paragraph_text), ' ')) AS word
# GROUP BY paragraph_number
# """)

# Display results from both views
display(spark.sql("SELECT * FROM workspace.bronze.vw_paragraph_word_count ORDER BY paragraph_number"))
#display(spark.sql("SELECT * FROM workspace.bronze.mv_paragraph_word_count ORDER BY paragraph_number"))
```
