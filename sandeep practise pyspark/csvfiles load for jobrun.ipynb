```python
# Path to the folder containing CSV files
folder_path = "/Volumes/workspace/jobs_schema/jobs_csv_files/four_csv_same_schema/"

# Read all CSV files from the folder into a single DataFrame
df = spark.read.option("header", "true").csv(f"{folder_path}/*.csv")

display(df)
```

```python
grouped_df = df.groupBy("city").count()
display(grouped_df)
```
